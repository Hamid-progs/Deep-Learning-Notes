```python
import pandas as pd
import numpy as np 

import tensorflow as tf
import keras
from tensorflow.keras.preprocessing.text import Tokenizer
from keras.utils import pad_sequences
```

```python
df = pd.read_csv('/kaggle/input/datasets/jp797498e/twitter-entity-sentiment-analysis/twitter_training.csv')
```

```python
df = df.iloc[:,2:]
df = df.rename(columns={'im getting on borderlands and i will murder you all ,': 'text',"Positive":"Sentiment"})
df.head()
```

```python
df.isna().sum()
```

```python
df.dropna(subset=['text'],inplace=True)
df.isna().sum()
```

```python
df.info()
```

```python
tokenizer = Tokenizer(oov_token='twitter')
```

```python
tokenizer.fit_on_texts(df['text'])
```

```python
# vocabulary
list(tokenizer.word_index.items())[:10]
```

```python
# frequency
list(tokenizer.word_counts.items())[:10]
```

```python
# number of rows
tokenizer.document_count
```

```python
sequences = tokenizer.texts_to_sequences(df['text'])
sequences[:10]
```

```python
# adding padding for making all vectors of same length
sequences = pad_sequences(sequences,padding='post')
sequences[:5]
```
